```python
from pathlib import Path
import pandas as pd

result_dir = Path("/home/song-z/Project/fm_evaluation/data/result/eegnet_8_30_128_v1")
files = sorted(result_dir.glob("*_sub-*.csv"))
if not files:
    raise FileNotFoundError("No EEGNet result files found.")
results = pd.concat(
    [pd.read_csv(file) for file in files],
    ignore_index=True,
)
results.to_csv("/home/song-z/Project/fm_evaluation/data/result/eegnet_all_8_30_128_v1.csv", index=False)
```
