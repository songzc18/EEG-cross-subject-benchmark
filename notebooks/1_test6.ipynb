```python
from fm_evaluation.modules.evaluation import evaluate_cv
from fm_evaluation.modules.utils import summarize
```

```python
filepath = "fm_evaluation/data/processed"#_8_30_128"
modelpath = "fm_evaluation/data/model_loso"
sub = 1
model = "eegnet"
train_sess = (1,)
train_phases = ("offline",)
protocol_sess = (1,)
protocol_phases = ("offline",)
test_sess = (1,2)
test_phases = ("offline","online")
n_splits = 5
folds = (0,)
k_values = (8,)
seed = 0
protocol_seeds = (0,)
subjects = (1,2,3)

model_kwargs = {
    "sfreq": 200,
    "max_epochs": 100,
    "lr": 0.001,
    "batch_size": 16,
    "module__kernel_length": 100,
    "module__drop_prob": 0.25,
}

ft_model="eegnet_full"
ft_kwargs = {
    "lr": 0.0001,
    "max_epochs": 5,
    "batch_size": 8,
}
```

```python
for protocol in ["SELF","POP","LOSO","CAL"]:
    val_size = 2 if protocol in ["SELF","POP"] else 1
    out_data = evaluate_cv(
        protocol = protocol,
        filepath=filepath,
        sub=sub,
        model=model,
        train_sess=train_sess,
        test_sess=test_sess,
        train_phases=train_phases,
        test_phases=test_phases,
        val_size=val_size,
        k_values=k_values,
        n_splits=n_splits,
        protocol_sess=protocol_sess,
        protocol_phases=protocol_phases,
        protocol_seeds=protocol_seeds,
        seed=seed,
        #subjects=subjects,
        model_kwargs=model_kwargs,
        ft_model=ft_model,
        ft_kwargs=ft_kwargs,
        modelpath=modelpath
    )

    print(protocol, out_data.shape)
    display(out_data.head())

    summarize(out_data)
```
